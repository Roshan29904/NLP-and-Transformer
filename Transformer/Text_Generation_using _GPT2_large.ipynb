```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

device
```

```python
model_name = 'openai-community/gpt2-large'

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name).to(device)

# Set pad_token for tokenizer and model config to suppress warnings during generation
tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.pad_token_id
```

```python
input_text = "i am really impressed"
```

```python
max_length = 128  #  this many tokens we want to generate
```

```python
# tokenize input
tokenized_input = tokenizer(input_text, return_tensors="pt")
```

```python
# we extract input_ids and attention_mask
input_ids = tokenized_input["input_ids"].to(device)
attention_mask = tokenized_input["attention_mask"].to(device)
```

```python
input_ids
```

### beam search


1.   if we run the model mutiple time we will observe that same text is bing generated




```python
output = model.generate(input_ids, attention_mask=attention_mask, max_length=max_length, num_beams=5, do_sample=False, no_repeat_ngram_size=2)
```

```python
output
```

```python
decoded_output = tokenizer.decode(output[0], skip_special_tokens=True)
print(f"Input Text: {input_text}")
print(f"Generated Text: {decoded_output}")
```

### nucleus search

1.   we run run the model we will observe that the same text is not being generated




```python


output_2 = model.generate(input_ids, attention_mask=attention_mask,
                        max_length=max_length, do_sample=True,top_p = 0.9)

```

```python
decoded_output_2 = tokenizer.decode(output_2[0], skip_special_tokens=True)
print(f"Input Text: {input_text}")

print(f"Generated Text: {decoded_output_2}")
```
