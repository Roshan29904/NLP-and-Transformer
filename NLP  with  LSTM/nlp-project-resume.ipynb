# Data cleaning

```python
import re
from nltk.corpus import stopwords
import nltk
```

```python
nltk.download('stopwords')

STOP_WORDS = set(stopwords.words('english'))
STOP_WORDS.add('none')
```

```python
TEXT_COLS = [
    'skills',
    'responsibilities',
    'positions',
    'degree_names',
    'major_field_of_studies',
    'related_skils_in_job',
    'career_objective',
]
```

```python
def preprocess(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    if text.strip() == 'none':          
        return ""
    text = re.sub(r'http\S+|www\S+', '', text) # remove URLs
    text = re.sub(r'[^a-z\s]', ' ', text) # keep only letters
    text = re.sub(r'\s+', ' ', text).strip() # normalise whitespace
    # remove stop words
    tokens = text.split()
    tokens = [t for t in tokens if t not in STOP_WORDS and len(t) > 1]
    return ' '.join(tokens)


def build_resume_text(row):
    parts = [preprocess(row[col]) for col in TEXT_COLS if col in row.index]
    return ' '.join(p for p in parts if p)


df["resume_text"] = df.apply(build_resume_text, axis=1)
```

# Tokenization and Padding

```python

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# tokenization

tokenizer = Tokenizer(num_words=3700 , oov_token='<OOV>')  #num_words=vocab_size
tokenizer.fit_on_texts(X_train)
```

```python
# now after tokenizing the words we will  convert text into  sequence of  number

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)
X_val_seq = tokenizer.texts_to_sequences(X_val)
```

```python
# now we need to pad the sequence so that all the sequences  are of same length 

X_train_pad = pad_sequences(X_train_seq, maxlen=200, padding="post", truncating="post") # Maxlen == max token in resume

X_test_pad = pad_sequences(X_test_seq, maxlen=200, padding="post", truncating="post")

X_val_pad = pad_sequences(X_val_seq, maxlen=200, padding="post", truncating="post")
```

# Model

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, LSTM, Bidirectional, Dense, Dropout, SpatialDropout1D, BatchNormalization, GlobalMaxPooling1D

from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
```

```python
# model

def build_bilstm():
    inp = Input(shape=(200,), name='resume_input')

    x = Embedding(3700, 128, name="embedding")(inp)

    x = SpatialDropout1D(0.2)(x)
    
    x = Bidirectional(LSTM(128, return_sequences=True,dropout=0.2, recurrent_dropout=0.1),name='bilstm_1')(x)
    
    x = Bidirectional(LSTM(64, return_sequences=False,dropout=0.2, recurrent_dropout=0.1),name='bilstm_2')(x)
    
    x = BatchNormalization()(x)

    x = Dense(64, activation='relu', kernel_regularizer=l2(1e-4))(x)

    x = Dropout(0.4)(x)

    # output dense for categorical (job position)
    cat_out = Dense(28, activation='softmax', name='category_out')(x)

    # output dense for label_interview(0,1) from matched_score
    int_out = Dense(1, activation='sigmoid', name='interview_out')(x)

    # model 
    model = Model(inputs=inp, outputs=[cat_out, int_out], name='Resume_BiLSTM')

    return model
```

```python
# build the model

model = build_bilstm()


# compilig the model
model.compile(
    optimizer=Adam(1e-4),
    loss={
        'category_out' : 'sparse_categorical_crossentropy',
        'interview_out': 'binary_crossentropy'
    },
    loss_weights={
        'category_out' : 1.0,
        'interview_out': 0.5
    },
    metrics={
        'category_out' : 'accuracy',
        'interview_out': 'accuracy'
    }
)

model.summary()
```

# Training

```python
callbacks = [
    EarlyStopping(
        monitor='val_category_out_accuracy',
        mode="max",
        patience=4,
        restore_best_weights=True,
        
    ),
    ModelCheckpoint(
        filepath="best_model.keras",
        monitor='val_category_out_accuracy',
        mode="max",
        save_best_only=True,
        
       
    ),
    ReduceLROnPlateau(
        monitor='val_loss',
        mode="min",
        factor=0.5, 
        patience=3,
        min_lr=1e-6,
    )
]
```

```python
history = model.fit(
    X_train_pad,
    {'category_out': yc_train, 'interview_out': yi_train},
    validation_data=(
        X_val_pad,
        {'category_out': yc_val, 'interview_out': yi_val}
    ),
    epochs=10,
    batch_size=32,
    callbacks=callbacks,
    
)
```

# Evaluation

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
cat_probs, int_probs = model.predict(X_test_pad, batch_size=64, verbose=1)

y_pred_cat = np.argmax(cat_probs, axis=1)
y_pred_int = (int_probs.squeeze() >= 0.5).astype(int)

cat_acc = accuracy_score(yc_test, y_pred_cat)
int_acc = accuracy_score(yi_test, y_pred_int)

```
